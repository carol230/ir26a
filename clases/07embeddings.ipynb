Un embedding (o incrustación) es fundamentalmente un truco matemático para que las computadoras entiendan el contexto y el significado humano. En términos técnicos, es una técnica de aprendizaje automático que mapea palabras, frases, imágenes o audio a vectores de números reales en un espacio de múltiples dimensiones.

A diferencia de enfoques tradicionales como la Bolsa de Palabras o TF-IDF (donde se crean vectores dispersos gigantescos basados en la simple frecuencia de aparición), los embeddings comprimen el significado en vectores densos de tamaño fijo. Esto significa que en lugar de tener un vector lleno de miles de ceros y unos pocos unos, tienes un vector más corto (típicamente entre 100 y 1024 dimensiones) lleno de números decimales que capturan matices semánticos.

```python
import gensim.downloader as api
# Cargar modelo
model = api.load("word2vec-google-news-300")
```

```python
vec = lambda word : model[word] # voy a agarrar vec, para tomar desde el modelo, el vector de cada palabra, y lo voy a usar para hacer la grafica

model ['dog'] # me da el vector de la palabra dog, que es un vector de 300 dimensiones

vec ("word")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_similarity([vec["fly"]], [vec("dog")])
```
